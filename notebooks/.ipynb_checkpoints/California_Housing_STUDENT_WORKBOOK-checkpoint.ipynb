# Part C — Protect the test set

## 11. Why stratify by income?

We want an **80% training / 20% test** split.

A normal random split is not automatically wrong. Here, however, income appears strongly related to house value, so we deliberately try to preserve the **broad income-group proportions** in both train and test sets.

### Important distinction
The original groups do **not** need equal numbers of observations.

Stratification means:

> preserve the proportion of each chosen group in train and test.

## 13. Create temporary income strata

We will use these broad groups:

- 0–1.5 → below about $15k
- 1.5–3.0 → about $15k–$30k
- 3.0–4.5 → about $30k–$45k
- 4.5–6.0 → about $45k–$60k
- 6.0+ → about $60k+

### Why this design?
It is a practical compromise:

- reasonably meaningful income widths in the dense part of the distribution;
- a combined upper tail instead of many small high-income groups;
- enough observations in each group.

These cut points are a **design choice**, not something uniquely determined by quartiles.

```python
# TODO 11
housing["income_cat"] = pd.cut(
    housing["median_income"],
    bins=[0., 1.5, 3.0, 4.5, 6.0, np.inf],
    labels=[1, 2, 3, 4, 5]
)

income_counts = housing["income_cat"].value_counts().sort_index()
income_percent = (
    housing["income_cat"]
    .value_counts(normalize=True)
    .sort_index() * 100
)

pd.DataFrame({
    "Count": income_counts,
    "Percent": income_percent.round(1)
})
```

## DEFINE → LEARN → APPLY

Before using the Pipeline, separate these three ideas:

- **DEFINE**: `Pipeline(...)`, `ColumnTransformer(...)`, `SimpleImputer(...)` describe the recipe. No values have been learned yet.
- **LEARN**: `model.fit(X_train, y_train)` learns preprocessing values **and** model parameters from the training data.
- **APPLY**: `model.predict(X_test)` applies the training-learned preparation to `X_test` and then predicts.

> **TRAIN teaches. TEST checks.**

## 18. Before Pipeline: `fit()` vs `transform()`

**`fit()` = LEARN from training data**

Examples:

- imputer learns the training median;
- scaler learns the training mean and spread;
- encoder learns the training categories.

**`transform()` = APPLY what was learned**

Examples:

- fill missing values using the learned median;
- scale values using the learned training parameters;
- create the learned one-hot columns.

### Rule

**TRAIN → learn + apply**  
**TEST / new data → apply only**

The test set must not teach us a new median, scale, or set of categories.

## 19. Why use a Pipeline?

For numerical columns we already know the required steps:

`missing value → Imputer → Scaler`

We could call these steps separately. A `Pipeline` keeps them together and applies them in the same order every time.

> **Pipeline = order of steps**

The code below is complete. Focus on reading it as a sequence rather than memorizing the syntax.

```python
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler

numeric_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

numeric_pipeline
```

## 20. A different path for categorical data

`ocean_proximity` contains labels rather than numerical measurements.

Its preparation path is:

`ocean_proximity → OneHotEncoder`

`handle_unknown="ignore"` allows later data to contain an unseen category without causing an error.

```python
from sklearn.preprocessing import OneHotEncoder

categorical_pipeline = Pipeline([
    ("onehot", OneHotEncoder(handle_unknown="ignore"))
])

categorical_pipeline
```

## 21. ColumnTransformer: send each column type to the correct path

- **Pipeline** = order of steps.
- **ColumnTransformer** = route columns to different paths.

Numerical columns:

`Imputer → Scaler`

Categorical columns:

`OneHotEncoder`

The prepared features are then joined for the model.

### Question

In the code below, which list should be used for the `"num"` route and which list for the `"cat"` route?

```python
# TODO 19
from sklearn.compose import ColumnTransformer

preprocess = ColumnTransformer([
    ("num", numeric_pipeline, __________________),
    ("cat", categorical_pipeline, __________________)
])

preprocess
```

- R² = 0 → no better, in squared-error terms, than predicting the mean target
