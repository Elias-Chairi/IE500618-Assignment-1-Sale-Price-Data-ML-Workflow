# Exploratory Data Analysis

## Step 1 - Explore and understand the data

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("../data/raw/AmesHousing.csv")

df.info()
```

```python
plt.hist(df["SalePrice"], bins = 30)
```

```python
na_sum = df.isna().sum()

na_sum[na_sum > 0]
```

# FUNN

VI har funnet ut at kolonnene har masse null verdier

```python
corr = df.corr(numeric_only=True)

target_corr = (
    corr["SalePrice"]
    .sort_values(ascending=True)
    .round(2)
)

target_corr[abs(target_corr) > 0.5]
```

```python
df.hist(figsize = (100, 70), bins = 30)
```
